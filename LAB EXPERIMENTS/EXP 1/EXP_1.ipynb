```python
import csv

def find_s(filename):
    with open(filename) as f:
        data = list(csv.reader(f))
        hypothesis = ['0'] * (len(data[0]) - 1)
        for row in data:
            if row[-1].strip().lower() == 'yes':
                for i in range(len(hypothesis)):
                    if hypothesis[i] == '0':
                        hypothesis[i] = row[i]
                    elif hypothesis[i] != row[i]:
                        hypothesis[i] = '?'
        return hypothesis

print("Most specific hypothesis:", find_s('enjoysport.csv'))
```
