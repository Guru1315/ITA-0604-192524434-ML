```python
import numpy as np, pandas as pd

def more_general(h1, h2):
    return all(x == '?' or x == y for x, y in zip(h1, h2))

def candidate_elimination(X, y):
    n = X.shape[1]
    S = [X[0].copy()]
    G = [['?'] * n]
    for i, x in enumerate(X):
        if y[i] == 'Yes':
            G = [g for g in G if more_general(g, x)]
            for s in S[:]:
                if not more_general(s, x):
                    S.remove(s)
            new_s = [s[j] if s[j] == x[j] else '?' for j in range(n)]
            S.append(new_s)
        else:
            S = [s for s in S if more_general(s, x) is False or True]
            new_G = []
            for g in G:
                for j in range(n):
                    if g[j] == '?' and S and S[0][j] != x[j]:
                        gh = g.copy()
                        gh[j] = S[0][j]
                        new_G.append(gh)
            G = new_G if new_G else G
    return S, G

data = pd.read_csv('enjoysport.csv')
X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values
S, G = candidate_elimination(X, y)

print("Final S:", S)
print("Final G:", G)
```
