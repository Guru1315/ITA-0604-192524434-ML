```python

# Credit Score Classification

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# 1. Create sample dataset
data = {
    'Income': [20000, 25000, 30000, 35000, 40000,
               45000, 50000, 60000, 70000, 80000],
    'Debt': [18000, 20000, 22000, 20000, 18000,
             15000, 12000, 10000, 8000, 5000],
    'Payment_Delay': [8, 7, 6, 5, 4, 3, 2, 1, 0, 0],
    'Credit_Score': ['Poor', 'Poor', 'Poor', 'Standard',
                     'Standard', 'Standard', 'Standard',
                     'Good', 'Good', 'Good']
}

df = pd.DataFrame(data)

# 2. Separate features and target
X = df[['Income', 'Debt', 'Payment_Delay']]
y = df['Credit_Score']

# 3. Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# 4. Create and train the model
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

# 5. Predict credit scores
y_pred = model.predict(X_test)

# 6. Display results
print("Actual Credit Scores:", list(y_test))
print("Predicted Credit Scores:", list(y_pred))

# 7. Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print("\nAccuracy:", round(accuracy * 100, 2), "%")

# 8. Display confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(
    y_test, y_pred, labels=['Poor', 'Standard', 'Good']
))

# 9. Predict score for a new customer
new_customer = pd.DataFrame(
    [[55000, 12000, 2]],
    columns=['Income', 'Debt', 'Payment_Delay']
)

prediction = model.predict(new_customer)
print("\nPredicted Credit Score:", prediction[0])
```
