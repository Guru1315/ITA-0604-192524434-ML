```python

import pandas as pd
import csv

# Create training data and save it as a CSV file
data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

columns = ['Sky', 'AirTemp', 'Humidity', 'Wind',
           'Water', 'Forecast', 'EnjoySport']

df = pd.DataFrame(data, columns=columns)
df.to_csv('training_data.csv', index=False)

# Read the CSV file
df = pd.read_csv('training_data.csv')
print("Training Data:")
print(df)

# Candidate-Elimination algorithm
X = df.iloc[:, :-1].values.tolist()
Y = df.iloc[:, -1].values.tolist()

S = ['0'] * len(X[0])
G = [['?'] * len(X[0])]

def covers(h, x):
    return all(a == '?' or a == b for a, b in zip(h, x))

def more_general(h1, h2):
    return all(a == '?' or a == b for a, b in zip(h1, h2))

for x, label in zip(X, Y):
    if label == 'Yes':
        G = [g for g in G if covers(g, x)]

        for i in range(len(S)):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

        G = [g for g in G if more_general(g, S)]

    else:
        new_G = []
        for g in G:
            if covers(g, x):
                for i in range(len(g)):
                    if g[i] == '?':
                        for value in df.iloc[:, i].unique():
                            if value != x[i] and (
                                S[i] == '?' or S[i] == value
                                or S[i] == '0'
                            ):
                                h = g.copy()
                                h[i] = value
                                if more_general(h, S):
                                    new_G.append(h)
            else:
                new_G.append(g)

        G = []
        for h in new_G:
            if h not in G:
                G.append(h)

print("\nFinal Specific Boundary (S):")
print(S)

print("\nFinal General Boundary (G):")
for g in G:
    print(g)

print("\nConsistent hypotheses are bounded by S and G.")
```
