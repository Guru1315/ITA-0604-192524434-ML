```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Generate sample data
np.random.seed(42)

X1 = np.random.normal(5, 1, 100)
X2 = np.random.normal(10, 1.5, 100)

X = np.concatenate([X1, X2]).reshape(-1, 1)

# Apply EM algorithm using Gaussian Mixture Model
model = GaussianMixture(n_components=2, random_state=42)
model.fit(X)

# Predict clusters
labels = model.predict(X)

# Display results
print("Cluster Means:")
print(model.means_.flatten().round(2))

print("\nCluster Variances:")
print(model.covariances_.flatten().round(2))

print("\nMixing Weights:")
print(model.weights_.round(2))

print("\nFirst 10 Cluster Labels:")
print(labels[:10])

# Plot clusters
plt.scatter(X[:, 0], np.zeros(len(X)), c=labels, cmap='viridis')
plt.scatter(model.means_.flatten(), [0, 0],
            color='red', marker='X', s=200, label='Cluster Means')

plt.title("Expectation-Maximization Algorithm")
plt.xlabel("Data Values")
plt.yticks([])
plt.legend()
plt.show()
```
