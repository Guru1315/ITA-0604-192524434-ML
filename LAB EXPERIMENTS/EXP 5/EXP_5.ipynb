```python
import numpy as np

def perceptron_train(X, y, epochs=20, lr=0.1):
    w = np.zeros(X.shape[1] + 1)
    for _ in range(epochs):
        for xi, target in zip(X, y):
            xi_b = np.insert(xi, 0, 1)
            out = 1 if np.dot(w, xi_b) > 0 else 0
            w += lr * (target - out) * xi_b
    return w

X = np.array([[0,0],[0,1],[1,0],[1,1]])
y_and = np.array([0,0,0,1])
w = perceptron_train(X, y_and)
print("Learned weights (AND):", w)
for xi in X:
    xi_b = np.insert(xi, 0, 1)
    print(xi, "->", 1 if np.dot(w, xi_b) > 0 else 0)
```
