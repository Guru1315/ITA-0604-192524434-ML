```python

import pandas as pd

data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Strong'],
    'Water': ['Warm', 'Warm', 'Warm', 'Warm'],
    'Forecast': ['Same', 'Same', 'Change', 'Same'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

print("Training Dataset:")
print(df)
print()

attributes = list(df.columns[:-1])
target = df.columns[-1]

hypothesis = ['0'] * len(attributes)

for index, row in df.iterrows():
    if row[target] == 'Yes':
        for i, attribute in enumerate(attributes):
            if hypothesis[i] == '0':
                hypothesis[i] = row[attribute]
            elif hypothesis[i] != row[attribute]:
                hypothesis[i] = '?'

        print(f"After example {index + 1}: {hypothesis}")

print("\nMost Specific Hypothesis:")
print(hypothesis)
```
